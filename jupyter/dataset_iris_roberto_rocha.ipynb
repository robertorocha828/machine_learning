```python
# Cargar datos
iris = load_iris()
X = iris.data
y = iris.target

#Configuracionj de Matplotlib
%matplotlib inline

# Reduccion PCA
X_reduce = PCA(n_components=3).fit_transform(X)

# Figura 3D
fig = plt.figure(figsize=(8,6))
ax = fig.add_subplot(111, projection='3d')

ax.scatter(X_reduce[:,0], X_reduce[:,1], X_reduce[:,2],
           c=y, cmap=plt.cm.Set1, s=40)
ax.set_title("First three PCA directions")
ax.set_xlabel("1st eigencevtor")
ax.set_ylabel("2nd eigenvetor")
ax.set_zlabel("3rd eignervetor")

plt.show()
```

```python
import matplotlib.pyplot as plt

x1_sepal_length = X[:, 0]
x2_sepal_width = X[:, 1]
x3_petal_length = X[:, 2]
x4_petal_width = X[:, 3]

%matplotlib inline
plt.figure(figsize=(15, 15))
plt.subplot(3, 3, 1)
plt.scatter(x1_sepal_length, x2_sepal_width, edgecolor='k', c=y)
plt.xlabel("Sepal Length (cm)")
plt.ylabel("Sepal Width (cm)")

plt.subplot(3, 3, 2)
plt.scatter(x1_sepal_length, x3_petal_length, edgecolor='k', c=y)
plt.xlabel("Sepal Length (cm)")
plt.ylabel("Petal Length (cm)")

plt.subplot(3, 3, 3)
plt.scatter(x1_sepal_length, x4_petal_width, edgecolor='k', c=y)
plt.xlabel("Sepal Length (cm)")
plt.ylabel("Petal Width (cm)")

plt.subplot(3, 3, 4)
plt.scatter(x2_sepal_width, x3_petal_length, edgecolor='k', c=y)
plt.xlabel("Sepal width (cm)")
plt.ylabel("Petal Length (cm)")

plt.subplot(3, 3, 5)
plt.scatter(x2_sepal_width, x4_petal_width, edgecolor='k', c=y)
plt.xlabel("Sepal width (cm)")
plt.ylabel("Petal Width (cm)")

plt.subplot(3, 3, 6)
plt.scatter(x3_petal_length, x4_petal_width, edgecolor='k', c=y)
plt.xlabel("Petal Length (cm)")
plt.ylabel("Petal Width (cm)")

plt.tight_layout() ## Ajusta el plot para que no se ensime
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Carga dataset iris
iris = sns.load_dataset("iris")

iris.head()
```

```python
plt.figure(figsize=(8,6))
sns.scatterplot(
    data =iris,
    x="sepal_length",
    y="sepal_width"
)

plt.title("Diagrama de dispersion: Sepal length vs Sepal Width")
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.scatterplot(
    data =iris,
    x="petal_length",
    y="petal_width",
    hue="species",
)

plt.title("Diagrama de dispersion: Sepal length vs Sepal Width")
plt.show()
```

```python
iris.info()
```

```python
plt.figure(figsize=(8,6))
sns.scatterplot(
    data =iris,
    x="sepal_length",
    y="petal_length",
    hue="species",
    size="sepal_width",
    sizes=(20,200)
)

plt.title("Diagrama de dispersion con tamaño de puntos segun sepal width")
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.scatterplot(
    data =iris,
    x="sepal_width",
    y="petal_width",
    hue="species",
    style="species"
)

plt.title("Diagrama de dispersion stilo con tamaño ")
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.scatterplot(
    data =iris,
    x="sepal_width",
    y="petal_width",
    hue="species",
    style="species",
    size="sepal_width",
    sizes=(20,300),
    alpha=0.7
)

plt.title("Diagrama con estilos y tamaños segun Sepal Width")
plt.show()
```
