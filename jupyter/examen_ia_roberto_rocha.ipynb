```python
edad_x = [20,25,30,35,40,45,50,55,60,65]
costo_consulta_y = [15,20,25,30,35,40,45,50,55,60]
costo_emergencia_y = [50,70,90,120,150,180,210,250,300,350]
costo_hospitalizacion_y = [200,300,400,550,700,900,1100,1400,1700,2000]
```

##### Pregunta 6: 
Se requiere generar una gráfica de línea que represente la relación entre la edad de los pacientes y el costo correspondiente de sus consultas médicas. Para ello, debes utilizar las variables:
edad_x: que contiene los diferentes rangos o valores de edad
costo_consulta_y: que contiene los costos asociados a cada grupo de edad
Requisitos específicos para la gráfica:
1. Eje X: Debe etiquetarse como "Edad (años)" y representar la variable edad_x
2. Eje Y: Debe etiquetarse como "Costo de Consulta (USD)" y representar la variable costo_consulta_y
3. Título: La gráfica debe llevar como título principal: "Relación entre Edad y Costo de Consulta Médica"
4. Leyenda: Incluir una leyenda que identifique la línea como "Consulta Médica"
5. Línea de tendencia: Mostrar claramente la progresión del costo según aumenta la edad

```python
plt.plot(edad_x,costo_consulta_y, label = 'Consulta Médica' )
plt.title('Relación entre Edad y Costo de Consulta Médica')
plt.xlabel('Edad (años)')
plt.ylabel('Costo de Consulta (USD)')
plt.legend()
plt.grid(True)
plt.show()
```

##### Pregunta 7:
Se requiere generar una gráfica comparativa de líneas que represente la relación entre la edad de los pacientes y los costos de dos tipos diferentes de servicios médicos. Para ello, debes utilizar las siguientes variables:
edad_x: que contiene los diferentes rangos o valores de edad
costo_consulta_y: que contiene los costos de consultas médicas regulares
costo_emergencia_y: que contiene los costos de servicios de emergencia médica
Requisitos específicos para la gráfica comparativa:
1. Primera línea (Consulta Médica):
Representar edad_x vs costo_consulta_y
Etiquetar como "Consulta Médica" en la leyenda
Utilizar el estilo de línea predeterminado
2. Segunda línea (Emergencia Médica):
Representar edad_x vs costo_emergencia_y
Etiquetar como "Emergencia Médica" en la leyenda
Configurar con las siguientes propiedades:
Grosor de línea: 3 puntos
Marcador: puntos sólidos (.)
Tamaño de marcador: 12
Estilo de línea: segmentada (--)
Color: rojo intenso (#d62728)
3. Elementos de la gráfica:
Eje X: Etiquetado como "Edad"
Eje Y: Etiquetado como "Costo ($)"
Título: "Comparación de Costos Médicos por Edad"
Leyenda: Mostrar ambas series claramente diferenciadas


```python
plt.plot(edad_x,costo_consulta_y, label = 'Consulta Médica' )
plt.plot(edad_x,costo_emergencia_y, label = 'Emergencia Médica', linewidth=3, marker = '.', markersize = 12, linestyle = '--', color = '#d62728' )
plt.title('Comparación de Costos Médicos por Edad')
plt.xlabel('Edad (años)')
plt.ylabel('Costo ($)')
plt.legend()
plt.grid(True)
plt.show()
```

##### Pregunta 8:
