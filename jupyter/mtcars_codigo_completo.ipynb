```python
# Cargar datos
df = sm.datasets.get_rdataset("mtcars").data
X = df.drop(columns="cyl").values
y = df["cyl"].astype("category").cat.codes
#Configuracionj de Matplotlib
%matplotlib inline

# Reduccion PCA
X_reduce = PCA(n_components=3).fit_transform(X)

# Figura 3D
fig = plt.figure(figsize=(8,6))
ax = fig.add_subplot(111, projection='3d')

ax.scatter(X_reduce[:,0], X_reduce[:,1], X_reduce[:,2],
           c=y, cmap=plt.cm.Set1, s=40)
ax.set_title("First three PCA directions")
ax.set_xlabel("1st eigencevtor")
ax.set_ylabel("2nd eigenvetor")
ax.set_zlabel("3rd eignervetor")

plt.show()
```

```python
import matplotlib.pyplot as plt

x1_mpg = df["mpg"]
x2_disp = df["disp"]
x3_hp = df["hp"]
x4_wt = df["wt"]

%matplotlib inline
plt.figure(figsize=(15, 15))
plt.subplot(3, 3, 1)
plt.scatter(x1_mpg, x2_disp, edgecolor='k', c=y)
plt.xlabel("MPG")
plt.ylabel("Displacement")

plt.subplot(3, 3, 2)
plt.scatter(x1_mpg, x3_hp, edgecolor='k', c=y)
plt.xlabel("MPG")
plt.ylabel("Horsepower")

plt.subplot(3, 3, 3)
plt.scatter(x1_mpg, x4_wt, edgecolor='k', c=y)
plt.xlabel("MPG")
plt.ylabel("Weight")

plt.subplot(3, 3, 4)
plt.scatter(x2_disp, x3_hp, edgecolor='k', c=y)
plt.xlabel("Displacement")
plt.ylabel("Horsepower")

plt.subplot(3, 3, 5)
plt.scatter(x2_disp, x4_wt, edgecolor='k', c=y)
plt.xlabel("Displacement")
plt.ylabel("Weight")

plt.subplot(3, 3, 6)
plt.scatter(x3_hp, x4_wt, edgecolor='k', c=y)
plt.xlabel("Horsepower")
plt.ylabel("Weight")

plt.tight_layout() ## Ajusta el plot para que no se ensime
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Carga dataset mtcars
mtcars = sm.datasets.get_rdataset("mtcars").data
mtcars["clase"] = mtcars["cyl"].astype(str) + " cilindros"

mtcars.head()
```

```python
plt.figure(figsize=(8,6))
sns.scatterplot(
    data =mtcars,
    x="wt",
    y="mpg"
)

plt.title("Diagrama de dispersion: Weight vs MPG")
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.scatterplot(
    data =mtcars,
    x="disp",
    y="hp",
    hue="clase",
)

plt.title("Diagrama de dispersion: Displacement vs Horsepower")
plt.show()
```

```python
mtcars.info()
```

```python
plt.figure(figsize=(8,6))
sns.scatterplot(
    data =mtcars,
    x="wt",
    y="mpg",
    hue="clase",
    size="hp",
    sizes=(20,200)
)

plt.title("Diagrama de dispersion con tamaño de puntos segun hp")
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.scatterplot(
    data =mtcars,
    x="disp",
    y="hp",
    hue="clase",
    style="clase"
)

plt.title("Diagrama de dispersion estilo segun cilindros")
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.scatterplot(
    data =mtcars,
    x="disp",
    y="hp",
    hue="clase",
    style="clase",
    size="disp",
    sizes=(20,300),
    alpha=0.7
)

plt.title("Diagrama con estilos y tamaños segun Displacement")
plt.show()
```
